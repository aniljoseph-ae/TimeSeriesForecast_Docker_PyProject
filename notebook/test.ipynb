```python
import yfinance as yf
import pandas as pd

def load_finance_ts(
    ticker: str,
    start_date: str,
    end_date: str,
) -> pd.DataFrame | None:
    try:
        # ---- Clean inputs ----
        ticker = ticker.strip().upper()
        start_date = str(start_date).strip()
        end_date = str(end_date).strip()

        # ---- Download ----
        data = yf.download(
            tickers=ticker,
            start=start_date,
            end=end_date,
            auto_adjust=False,
            progress=False,
        )

        # ---- Guard: empty download ----
        if data.empty:
            raise ValueError("No data returned. Check ticker or date range.")

        # ---- Handle MultiIndex columns ----
        if isinstance(data.columns, pd.MultiIndex):
            data.columns = data.columns.droplevel(1)

        # ---- Remove column index name (e.g. 'Price') ----
        data.columns.name = None

        # ---- Time series formatting ----
        data.index.name = "date"
        data = data.sort_index()

        return data

    except Exception as e:
        print(f"Error loading finance data: {e}")
        return None
```

```python
import pandas as pd
import yfinance as yf


def load_finance_time_series(
    ticker: str,
    start_date: str,
    end_date: str,
) -> pd.DataFrame:
    """
    Load financial time-series data and normalize it into a clean,
    forecasting-ready DataFrame.

    Returns:
        DataFrame indexed by date with columns:
        ['Open', 'High', 'Low', 'Close', 'Volume', 'Adj Close' (optional)]
    """

    # --------------------------------------------------
    # 1️⃣ Sanitize inputs (VERY IMPORTANT)
    # --------------------------------------------------
    ticker = ticker.strip().upper()
    start_date = str(start_date).strip()
    end_date = str(end_date).strip()

    # --------------------------------------------------
    # 2️⃣ Download data
    # --------------------------------------------------
    data = yf.download(
        tickers=ticker,
        start=start_date,
        end=end_date,
        auto_adjust=False,   # explicit to avoid future warning
        progress=False,
    )

    if data.empty:
        raise ValueError("No data returned. Check ticker or date range.")

    # --------------------------------------------------
    # 3️⃣ FIX: MultiIndex on COLUMNS (ticker level)
    # --------------------------------------------------
    if isinstance(data.columns, pd.MultiIndex):
        # drop ticker level (AAPL)
        data.columns = data.columns.droplevel(1)

    # remove column index name like "Price"
    data.columns.name = None

    # --------------------------------------------------
    # 4️⃣ FIX: MultiIndex on INDEX (Date, Price)
    # --------------------------------------------------
    if isinstance(data.index, pd.MultiIndex):
        if "Price" in data.index.names:
            # unstack price into columns
            data = data.unstack(level="Price")
            data.columns = data.columns.droplevel(0)

    # --------------------------------------------------
    # 5️⃣ Final cleanup
    # --------------------------------------------------
    data.index = pd.to_datetime(data.index)
    data.index.name = "date"
    data = data.sort_index()

    return data
```
